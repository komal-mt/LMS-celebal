# Load the Dataset

```python
import pandas as pd

df = pd.read_csv("./student-scores.csv")
df.head()
```

```python
df.drop("career_aspiration",axis=1, inplace=True)
df.head()
```

```python
# Drop all columns with object (string) dtype
df.drop(["first_name","last_name","email"],axis = 1,inplace = True)
df.head()
```

# Data Preprocessing

```python
df.head()
```

```python
# now we need to convert string values to the nummeric like male = 1 , female = 0, True = 1 , False = 0
```

```python
df['gender'] = df['gender'].replace({'male': 1, 'female': 0})
df.head()
```

```python
df['part_time_job'] = df['part_time_job'].replace({True: 1, False: 0})
df['extracurricular_activities'] = df['extracurricular_activities'].replace({True: 1, False: 0})
df.drop("id",inplace = True,axis =1)
```

```python
df.head()
```

```python
# now add all subject colums to one 

df['total_marks'] = df[['math_score', 'history_score', 'english_score','physics_score','chemistry_score','biology_score','geography_score']].sum(axis=1)
df.head()
```

```python
df.drop(['math_score', 'history_score', 'english_score','physics_score','chemistry_score','biology_score','geography_score'],axis = 1,inplace = True)
df.head()
```

```python
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
```

# Featuer extracting

```python
from sklearn.preprocessing import StandardScaler

features = df[['gender','part_time_job','absence_days','extracurricular_activities','weekly_self_study_hours']]
target = df['total_marks']
scaler = StandardScaler()
x_scaled = scaler.fit_transform(features)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x_scaled, target, test_size=0.2, random_state=42)
```

# model buliding

## linear regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

lr = LinearRegression()
lr.fit(x_train, y_train)

y_pred = lr.predict(x_test)
print("linear regression R2 score:", r2_score(y_test, y_pred))
print("MSE:", mean_squared_error(y_test, y_pred))
```

## random forset

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor()
rf.fit(x_train, y_train)
rf_pred = rf.predict(x_test)

print("random forest R2 Score:", r2_score(y_test, rf_pred))
```

# feature importance

```python
importances = rf.feature_importances_
for feat, score in zip(features.columns, importances):
    print(f"{feat}: {score:.4f}")
```
